```python
from pathlib import Path
import numpy as np
import pandas as pd
from pprint import pprint
```

```python
orig_dir = Path('origdata')
s_files = [orig_dir/'spec_all_clean.csv',
           orig_dir/'spec_swir_clean.csv',
           orig_dir/'spec_vnir_clean.csv']
t_files = [orig_dir/'trait_clean.csv']

comp_dir = Path('compdata')
comp_dir.mkdir(parents=True, exist_ok=True)
```

```python
col_remap = {}
for sf in s_files:
    df = pd.read_csv(sf)
    for c in df.columns:
        try:
            f = float(c)
            col_remap[c] = f'X_{f:4.3f}'
        except:
            col_remap[c] = c
```

```python
tdf = pd.read_csv(t_files[0])
print(f'{tdf.shape = } (loaded)')

tdf = tdf.dropna(subset=['closest_spec_date'])
# print(f'{tdf.shape = } (dropped NA in closest_spec_date)')
# pprint(tdf['Comment'].value_counts())

tdf = tdf.drop_duplicates(subset=['Comment'])
print(f'{tdf.shape = } (dropped duplicates in Comment)')
# pprint(tdf['Comment'].value_counts())

tdf['sample_id'] = tdf.apply(lambda r: str(r['Comment']) + '_' + str(r['closest_spec_date']),
                             axis=1)
# pprint(tdf.head(2))
```

```python
sdf = pd.read_csv(s_files[-1])
print(f'{sdf.shape = } (loaded)')

sdf['d'] = sdf.apply(lambda r: str(r['flightline']).split('-')[0][-8:],
                     axis=1)
sdf['date'] = sdf.apply(lambda r: (str(r['d'])[:4] + '-' + 
                                   str(r['d'])[4:6] + '-' + 
                                   str(r['d'])[6:8]),
                        axis=1)
sdf['sample_id'] = sdf.apply(lambda r: str(r['Name']) + '_' + str(r['date']),
                             axis=1)
```

```python
tdf_sids = set(tdf['sample_id'].values)
print(f'{len(tdf_sids) = }')

sdf_sids = set(sdf['sample_id'].values)
print(f'{len(sdf_sids) = }')

sids = sdf_sids & tdf_sids
print(f'{len(sids) = }')
```
